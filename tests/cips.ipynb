```python
from corrosions import CIPS
import pandas as pd
```

```python
dirs = [
    r"D:\Data\CIPS_2024 (update)",
    r"D:\Data\CIPS FINAL 2025"
]
```

```python
results = []
for index, dir in enumerate(dirs):
    cips = CIPS(
        file_or_dir=dir,
        sheet_name='Data' if index == 1 else None,
        verbose=True,
        overwrite=True,
    )
    cips.normalize()
    results = results + cips.results
```

```python
faileds = []
for result in results:
    if not result['success']:
        faileds.append(result)
```
