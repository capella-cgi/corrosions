```python
import os
import pandas as pd
```

```python
acvg_dcvg_dir = os.path.join(os.getcwd(), 'acvg_dcvg')
os.makedirs(acvg_dcvg_dir, exist_ok=True)
```

```python
file = r"D:\Temp\Analisis Data IDDA-ECDA - 2024.xlsx"
```

```python
df = pd.read_excel(file, sheet_name='Survey ACVG - DCVG')
```

```python
df.head()
```

```python
segment_codes = df['segment_code'].unique().tolist()
```

```python
for segment_code in segment_codes:
    new_df = df[df['segment_code'] == segment_code]
    filename = f"acvg-dcvg-{segment_code}.xlsx"
    filepath = os.path.join(acvg_dcvg_dir, filename)
    new_df.to_excel(filepath, index=False)
    print(filepath)
```
