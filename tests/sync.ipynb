```python
from corrosions import Sync
import os
import pandas as pd
```

```python
normalize_acvg_dcvg_dir = r"D:\Projects\corrosions\tests\normalize\acvg_dcvg\excel"
normalize_cips_dir = r"D:\Projects\corrosions\tests\normalize\cips\excel"
normalize_pcm_dir = r"D:\Projects\corrosions\tests\normalize\pcm\excel"
```

```python
df = pd.read_excel(r"D:\Projects\corrosions\tests\IDDA - File List.xlsx")
```

```python
syncs: list[Sync] = []
for index, label in df.iterrows():
    acvg_dcvg_file = None if pd.isnull(label["acvg_dcvg_file"]) else label["acvg_dcvg_file"]
    pcm_file = label["pcm_file"]
    cips_file = label["cips_file"]

    sync = Sync(
        area=label["area_code"],
        segment_code=label["segment_code"],
        year=label["year"],
        pipe_diameter=label["pipe_diameter"],
        length=label["length"],
        normalized_acvg_dcvg_file=os.path.join(normalize_acvg_dcvg_dir, acvg_dcvg_file) if acvg_dcvg_file else None,
        normalized_cips_file=os.path.join(normalize_cips_dir, cips_file),
        normalized_pcm_file=os.path.join(normalize_pcm_dir, pcm_file),
        verbose=True,
    )
    syncs.append(sync)
```

```python
sync_list = []
for _sync in syncs:
    _sync.fix().recalculate_distance()
    sync_list.append(_sync.dict)
```
