```python
from corrosions import Sync, PCM
import os
import pandas as pd
```

```python
labels = r"D:\Projects\corrosions\tests\labels.xlsx"
```

```python
df = pd.read_excel(labels)
```

```python
df
```

```python
syncs: list[Sync] = []
for index, label in df.iterrows():
    acvg_dcvg_file = None if pd.isnull(label["normalized_acvg_dcvg_file"]) else label["normalized_acvg_dcvg_file"]
    pcm_file = label["normalized_pcm_file"]
    cips_file = label["normalized_cips_file"]

    print(index, pcm_file)

    sync = Sync(
        area=label["area"],
        segment_code=label["segment_code"],
        year=label["year"],
        pipe_diameter=label["pipe_diameter"],
        length=label["length"],
        normalized_acvg_dcvg_file=acvg_dcvg_file if acvg_dcvg_file else None,
        normalized_cips_file=cips_file,
        normalized_pcm_file=pcm_file,
        verbose=False,
    )
    syncs.append(sync)
```

```python
sync_list = []
for _sync in syncs:
    # print(_sync.normalized_pcm_file)
    _sync.fix().recalculate_distance()
    sync_list.append(_sync.dict)
```
