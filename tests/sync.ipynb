```python
labels = [
    {
        "area": "Tangerang",
        "year": 2024,
        "pipe_diameter": 16,
        "length": 16.35,
        "segment_code": "offtake-serpong-ms-batu-ceper-16",
        "files": {
            "acvg_dcvg": "acvg-dcvg-offtake-serpong-ms-batu-ceper-16-sheet1.xlsx",
            "pcm": "pcm-tgr-16-inch-offtake-serpong-ms-batu-ceper-15-65-km-sequential-files.xlsx",
            "cips": "cips-tgr-16-inch-offtake-serpong-ms-batu-ceper-survey-data.xlsx"
        }
    },
    {
        "area": "Bekasi",
        "year": 2024,
        "pipe_diameter": 24,
        "length": 13.59,
        "segment_code": "rawa-maju-tegal-gede-offtake-deltamas-24",
        "files": {
            "acvg_dcvg": "acvg-dcvg-rawa-maju-tegal-gede-offtake-deltamas-24-sheet1.xlsx",
            "pcm": "pcm-bks-24-inch-rawa-maju-tegal-gede-offtake-deltamas-sheet1.xlsx",
            "cips": "cips-bks-24-inch-rawa-maju-tegal-gede-offtake-surya-cipta-sheet1.xlsx"
        }
    },
    {
        "area": "Bogor",
        "year": 2024,
        "pipe_diameter": 16,
        "length": 16.35,
        "segment_code": "jonggol-cimanggis-16",
        "files": {
            "acvg_dcvg": "acvg-dcvg-jonggol-cimanggis-16-sheet1.xlsx",
            "pcm": "pcm-bgr-16-inch-jonggol-cimanggis-9-95-km-sequential-file.xlsx",
            "cips": "cips-bgr-16-inch-jonggol-cimanggis-9-95-km-survey-data.xlsx"
        }
    },
]
```

```python
syncs: list[Sync] = []
for label in labels:
    sync = Sync(
        area=label["area"],
        segment_code=label["segment_code"],
        year=label["year"],
        pipe_diameter=label["pipe_diameter"],
        length=label["length"],
        normalized_acvg_dcvg_file=os.path.join(normalize_acvg_dcvg_dir, label["files"]["acvg_dcvg"]),
        normalized_cips_file=os.path.join(normalize_cips_dir, label["files"]["cips"]),
        normalized_pcm_file=os.path.join(normalize_pcm_dir, label["files"]["pcm"]),
        verbose=True,
    )
    syncs.append(sync)
```
